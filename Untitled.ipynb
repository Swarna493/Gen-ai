```python
import numpy as np

# Activation functions
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

def tanh(x):
    return np.tanh(x)

def tanh_derivative(x):
    return 1 - np.tanh(x) ** 2

# Merge 2 input layers (arrays) into one input vector
input1 = np.array([[0.1, 0.2]])
input2 = np.array([[0.3, 0.4]])
X = np.concatenate((input1, input2), axis=1)  # Shape: (1, 4)

# Define layer sizes
input_size = 4
hidden1_size = 5
hidden2_size = 3
output_size = 1

# Random weight initialization
np.random.seed(42)
W1 = np.random.randn(input_size, hidden1_size)
b1 = np.zeros((1, hidden1_size))

W2 = np.random.randn(hidden1_size, hidden2_size)
b2 = np.zeros((1, hidden2_size))

W3 = np.random.randn(hidden2_size, output_size)
b3 = np.zeros((1, output_size))

# ====== FORWARD PASS with ACTIVATION FUNCTIONS ======

# Layer 1 → ReLU
z1 = np.dot(X, W1) + b1
a1 = relu(z1)

# Layer 2 → Tanh
z2 = np.dot(a1, W2) + b2
a2 = tanh(z2)

# Output layer → Sigmoid
z3 = np.dot(a2, W3) + b3
output = sigmoid(z3)

# Print activations at each layer
print("Input:", X)
print("After Hidden Layer 1 (ReLU):", a1)
print("After Hidden Layer 2 (Tanh):", a2)
print("Final Output (Sigmoid):", output)
```
